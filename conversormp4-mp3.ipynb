```python
import subprocess
import imageio_ffmpeg

FFMPEG = imageio_ffmpeg.get_ffmpeg_exe()  # caminho absoluto do ffmpeg.exe

def preparar_audio(origem: str, destino: str):
    # converte mp4/ogg do WhatsApp -> mp3 mono 48k, pronto para a API
    subprocess.run([
        FFMPEG, '-y', '-i', origem,
        '-vn', '-ac', '1', '-b:a', '48k',
        destino
    ], check=True)

preparar_audio(r'C:\Users\ia01\Desktop\5. Costumer Sucess\vira-ata\aud.mp4',
               r'C:\Users\ia01\Desktop\5. Costumer Sucess\vira-ata\aud.mp3')
```
