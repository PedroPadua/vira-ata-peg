```python
from dotenv import load_dotenv, find_dotenv
import openai
_= load_dotenv(find_dotenv())
client = openai.OpenAI()
```

```python
def chat_openai(messages, model='gpt-5.6-luna', effort='low', stream=False):
    resposta = client.chat.completions.create(
        model=model,
        messages=messages,
        reasoning_effort=effort, 
        stream=stream,
    )
    return resposta

mensagens = [{
    'role':'user',
    'content' : 'Qual a cor do céu?'
}]
resp = chat_openai(mensagens)
resp.choices[0].message.content
```
